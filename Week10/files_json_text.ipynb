### 👨‍🏫Concept-2: mode (w vs a vs r)

| Mode | Matlab | Dhyaan |
|---|---|---|
| `"r"` | Read (padho) | Default file na ho toh error |
| `"w"` | Write (likho) | Purana sab MITA deta hai! |
| `"a"` | Append (jodo) | End mein add kate hai, purana safe |
| "SABSE BADA trap: `"w"` mode file ka SAARA purana content MITA deta hai aur naya likhta hai. Agar purana data bachana hai aur naya jodna hai, `"a"` (append) use karo . Yeh galti sabko ek baar lagti hai." |

### 👨‍🏫 Concept-4: fiel ke saath JSON (dump/load)

```python
data = {
	"name": "Agentic AI",
	"age": 15,
	"Subjects":["Math", "Science"],
	"is_Student": True
}
with open("stud_record.json", "w") as f:
	json.dump(data, f, indent=4)
```

```python
with open("stud_record.json", "r") as f:
	data_from_file = json.load(f)
```

```python
data_from_file
```

```python
data_from_file["Subjects"]
```

*"`dump`/`load` (bina 's') seedhe FILE ke saath kaam karte hain. `dumps` / `loads`(with 's') STRING ke saath . Yeh farak yad rakho - yeh confuse karte hai. "*

| Function | kaam |
|---|---|
| `json.dumps(data)` | dict + JSON **string** |
| `json.loads(text)` | JSON string → dict |
| `json.dump(data, file)` | dict → JSON **file** | 
| `json.load(file)` | JSON file → dict | 

### 🏷️ MINI Projcet : Save Agent Chat History

```python
def create_history()->None:
	with open("agent_chat_history.json", "w") as f:
		json.dump([], f, indent=4)

def add_history(role: str, content: str)->None:
	
	with open("agent_chat_history.json", "r") as f:
		history = json.load(f)

	history.append({"role":role, "content": content})

	with open("agent_chat_history.json", "w") as f:
		json.dump(history, f, indent = 4)

def read_history()->list:
	with open("agent_chat_history.json", "r") as f:
		print(json.load(f))			
```

```python
create_history()
```

```python
add_history("user", "Hi, How are you")
```

```python
add_history("system", "Act as devloper")
```

```python
read_history()
```

# 🌈💥 CSV & Regex :

### 👨‍🏫 Concept-1: Comma-Seperated Value

```python
import csv

rows = [
			["Name", "Marks"],
			["Agentic", 87],
			["AI", 79]
]

with open("marks.csv","w", newline="",encoding="utf-8") as f:
	writer = csv.writer(f)
	writer.writerows(rows)
```

```python
with open("marks.csv", "r", encoding="utf-8") as f:
	reader = csv.reader(f)
	for row in reader:
		print(row)
```

### 👨‍🏫 Concept-2: CSV as Dictionaries (Zyada readable)

```python
with open("marks.csv", "r", encoding="utf-8") as f:
	reader = csv.DictReader(f)
	for row_dict in reader:
		print(row_dict["Name"], row_dict["Marks"])
```

### 👨‍🏫 Concept-3: Regex (Pattern dhoondhna)

```python
import re
```

```python
text = """My contact : email >> agentic_ai@gmail.com, 
			and my mobile no >> 1234567890, my secondary mo.no : 7037768022
"""
numbers = re.findall(r"\d+", text)
print(numbers)
```

```python
text = """My contact : email >> agentic_ai@gmail.com, 
			and my mobile no >> 1234567890, my secondary mo.no : 7037768022
"""
numbers = re.search(r"\d+", text)
print(numbers.group())
```

```python
list_words = text.split(" ")
```

```python
def get_email(text):
	list_words = text.split(" ")
	for word in list_words:
		if "@" in word:
			print(word)

get_email(text)
```

```python
email = re.search(r"\w+@\w+\.\w+", text)
print(email.group())
```

### 👨‍🏫 Concept-4: common regex patterns

```python
text = "I have 3 cats and 12 dogs. #pets #animals"
print(re.findall(r"\d+", text))
print(re.findall(r"\#w+", text))
print(re.sub(r"\d+", "N", text))
```
