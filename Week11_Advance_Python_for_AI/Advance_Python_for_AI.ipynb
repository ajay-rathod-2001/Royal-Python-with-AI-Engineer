### 👨‍🏫 Concept-2.5: Iterables, Iterators & Generators (dono mein fark)

* Iterable :- An Iterable is any Python object (like a list, tuple, or string ), that can be looped over in a `for` loop. It contains data in memory but does not track its own iteration state or current position . Passing Iterable to `iter(x)` create and returns a new Iterator object.

* Iterator :- An Iterator is a staeful stream object that remebers its current position dring iteration. It featches the next item one-at-a-time whenever `next(iterator)` is called. When no elements remain, calling `next()` raises a `StopIteration` exception.

* Generator :- A Generator is a special type of function uses `yield` to return values lazily one at time. Calling a generator fuction automatically returns a specialized Iterator object without storing all items in memory at once. It pause execution at `yield` and resumes seamlessly when the next item is requested, making  it ideal for LLM token streaming.

```python
numbers = [1,2,3,4,5,6]
```

```python
iter = iter(numbers)
```

```python
iter
```

```python
print(next(iter))
```

```python
print(next(iter))
```

## 🔀🏷️ Summery Table:

| Concept | Memory Behavior | State Tracking | How it is Created |
| :---| :---| :---| :---|
| **Iterable** | Stores all elements in memory at once | Does not tarck current position | defined as `[1,2,3]`, tuples, strings |
| **Iterator** | Fetches itema one at a time on demand | Remembers current position; advances via `next()` | Created via `iter(iterable)` | 
| **Generator** | Computers values lazily on demand | Pauses and resumes function state at `yield` | Function written with `yield` |

*"`Iterable` = sequence jisme items hain. `Iterator` = pointer/stream jo state track karat hai. Generator function calling  se ek **Iterator** object banta hai!."*

### 👨‍🏫 Concept-2: decorator ka basic dhaacha

```python
def decorator_function(func):
	def wrapper():
		print("before function")
		result = func()
		print("after function")
		return result
	return wrapper


def greet():
	return "Hello, World"

deco = decorator_function(greet)
print(deco())
```

### 👨‍🏫 Concept-3: @ syntax (sundar shortcut)

```python
@decorator_function
def greet():
	return "Hello, World"
```

```python
greet()
```

```python
def line_decorator(func):
	"""This is a line decorator that adds a line of equal sign before and after the function call."""
	def wrapper():
		print("="*20)
		result = func()
		print(result)
		print("="*20)

	return wrapper
```

```python
@line_decorator
def greet():
	return "How are you?"
```

```python
greet()
```

```python
@line_decorator
def rani():
	return "I Love my country"
rani()
```

### 👨‍🏫 Concept-3: argument wale functions ko wrap karna (*args, **kwargs)

```python
import functools
```

```python
def time(func):
	"""This is a decorators that calculation the time taken to executed a function."""

	@functools.wraps(func)
	def wrapper(*args, **kwargs):
		import time
		start_time = time.time()
		result = func(*args, **kwargs)
		end_time = time.time()
		print(f"Time taken execute function: {func.__name__} : {end_time - start_time} seconds.")
		print(f"The Function {func.__name__} use for {func.__doc__}")
		
		return result
	return wrapper 
```

```python
@time
def addition(a, b):
	"""This is the function thats ADDS two number"""
	return a + b
```

```python
addition(3,4)
```

```python
@time
def mult(a, b, c):
	"""This is the function thats MULT two number"""
	return a * b * c
mult(3,5,4)
```

```python
addition.__name__
```

```python
addition.__doc__
```

```python
mult.__doc__
```

### 👨‍🏫 Conecept-2: contextmanagers se apna banao

```python
from contextlib import contextmanager
import time
```

```python
@contextmanager
def timer(label):
	start_time = time.time()
	print(f"Starting {label}...")
	print("Staring time is", start_time)
	yield
	end_time = time.time()
	print(f"{label} took {end_time - start_time} seconds")
```

```python
with timer("my task"):
	total= sum(range(100))
	print("Total sum is", total)
```

### 👨‍🏫 Concept-6: Protocol (first pass - duck typing)

```python
from typing import Protocol
```

```python
class Runnable(Protocol):
	def run(self) -> str:...

def execute(tool: Runnable) -> str:
	return tool.run()

```

```python
class SearchTools:
	def run(self) -> str:
		return "Searching........."
```

```python
tool = SearchTools()

execute(tool)
```

```python
class SearchToolsWithoutRun:
	def ran(self) -> str:
		return "Searching........."
```

```python
bad_tool =  SearchToolsWithoutRun()
execute(bad_tool)
```
